```python
df["Age"].describe()
```

```python
df["AgeGroup"] = pd.cut(
    df["Age"],
    bins=[17, 24, 34, 44, 54, 60],
    labels=["18-24", "25-34", "35-44", "45-54", "55-60"]
)
```

```python
df["AgeGroup"].value_counts().sort_index()
```

```python
df.groupby("AgeGroup")["Attrition"].value_counts()
```

```python
age_attrition_rate = (
    df.groupby("AgeGroup")["Attrition"]
    .apply(lambda x: (x == "Yes").sum() / len(x) * 100)
)

age_attrition_rate
```

```python
df["MonthlyIncome"].describe()
```

```python
df["IncomeGroup"] = pd.cut(
    df["MonthlyIncome"],
    bins=[0, 3000, 6000, 10000, 15000, float("inf")],
    labels=["0-3K", "3K-6K", "6K-10K", "10K-15K", "15K+"]
)
```

```python
df["IncomeGroup"].value_counts().sort_index()
```

```python
income_attrition_rate = (
    df.groupby("IncomeGroup", observed=True)["Attrition"]
    .apply(lambda x: (x == "Yes").sum() / len(x) * 100)
)

income_attrition_rate
```

```python
df["YearsAtCompany"].describe()
```

```python
df["TenureGroup"] = pd.cut(
    df["YearsAtCompany"],
    bins=[-1, 2, 5, 10, 20, float("inf")],
    labels=["0-2", "3-5", "6-10", "11-20", "21+"]
)
```

```python
df["TenureGroup"].value_counts().sort_index()
```

```python
tenure_attrition_rate = (
    df.groupby("TenureGroup", observed=True)["Attrition"]
    .apply(lambda x: (x == "Yes").sum() / len(x) * 100)
)

tenure_attrition_rate
```
